# ***Análisis de Spotify - All Time Top 2000s Mega Dataset***
---

## Descripción General
En la era digital, la industria musical ha experimentado una transformación profunda impulsada por las plataformas de *streaming*. Spotify, líder del mercado global, procesa diariamente millones de reproducciones y genera un volumen masivo de datos sobre las características sonoras y el éxito de las canciones.

El dataset **Spotify - All Time Top 2000s Mega Dataset** reúne un catálogo masivo de las 2.000 canciones más destacadas de todos los tiempos en la plataforma. A través de métricas acústicas como el ritmo (BPM), la bailabilidad (*danceability*), la energía (*energy*), la positividad (*valence*) y la popularidad (*popularity*), este conjunto de datos permite analizar los factores que definen a los grandes éxitos musicales a lo largo de las décadas.

El presente caso de estudio aborda este mega catálogo para identificar patrones de consumo, la evolución de los atributos sonoros según el año de lanzamiento y los elementos clave que inciden de manera directa en el impacto de una producción musical.

---

## Objetivo del Análisis
El objetivo principal de este proyecto es explorar y analizar el **Spotify - All Time Top 2000s Mega Dataset** para responder a las siguientes preguntas clave de negocio:

* **Evolución Temporal:** ¿Cómo han variado las características sonoras (BPM, energía, bailabilidad) en las canciones más populares según su año de lanzamiento (`Year`)?
* **Factores de Popularidad:** ¿Existe una relación directa o correlación entre las métricas acústicas de un tema y su nivel de popularidad (`Popularity`)?
* **Distribución de Géneros y Artistas:** ¿Qué géneros (`Top Genre`) y artistas concentran la mayor cantidad de éxitos dentro de este Top 2000 y cómo se caracterizan sus perfiles sonoros?

```python
import pandas as pd
import csv

# Carga del dataset manejando comillas y delimitadores complejos
df = pd.read_csv(
    "Spotify-2000.csv",
    encoding='latin-1',
    quoting=csv.QUOTE_NONE,
    on_bad_lines='skip'
)

# Vistazo rápido
print("Dimensiones (filas, columnas):")
print(df.shape)

print("\nNombres de columnas:")
print(df.columns)

print("\nPrimeras 10 filas:")
display(df.head(10))

print("\nUltimas 5 filas:")
display(df.tail(5))

print("\nMuestra aleatoria de 5 filas:")
display(df.sample(5, random_state=42))

# Tipos de datos
print("\nTipos de datos y nulos:")
df.info()

print("\nEstadisticas descriptivas completas:")
display(df.describe(include='all'))
```

```python
# 1. Renombrar las cabeceras con nombres limpios
df = df.rename(columns={df.columns[0]: 'Index', 'Popularity;;': 'Popularity'})

# 2. Quitar puntos y comas, comillas y espacios en blanco de la columna Popularity
df['Popularity'] = df['Popularity'].astype(str).str.replace(';;', '', regex=False)
df['Popularity'] = df['Popularity'].str.replace('"', '', regex=False).str.strip()

# 3. Convertir a número (int) forzando errores si encuentra texto no numérico
df['Popularity'] = pd.to_numeric(df['Popularity'], errors='coerce')

# 4. Verificar que se haya convertido correctamente
print("Tipo de dato de Popularity:")
print(df['Popularity'].dtype)

print("\nResumen de Popularity:")
print(df['Popularity'].describe())
```

```python
# Comprobacion de estructura actual
print("Nombres de todas las columnas:")
print(df.columns.tolist())

print("\nVista previa de las primeras 5 filas (Index, Title, Artist, Popularity):")
display(df[['Index', 'Title', 'Artist', 'Popularity']].head())

print("\nInformacion general del DataFrame:")
df.info()
```

```python
import pandas as pd
import numpy as np

# VALORES FALTANTES (NaN)
print("--- VALORES FALTANTES (NaN) ---")
nulos = df.isna().sum().sort_values(ascending=False)
pct_nulos = (df.isna().mean() * 100).round(2)
tabla_nulos = pd.DataFrame({'nulos': nulos, '%_nulos': pct_nulos})
display(tabla_nulos.head(10))

# DUPLICADOS
print("\n--- DUPLICADOS ---")
dup_total = df.duplicated().sum()
print(f"Filas duplicadas encontradas: {dup_total}")

# Eliminar duplicados
df = df.drop_duplicates()

# DETECCIÓN DE OUTLIERS (IQR)
print("\n--- DETECCIÓN DE OUTLIERS (IQR) ---")

cols_num = [
    'beats_per_minute_bpm', 'energy', 'danceability', 'loudness_db',
    'liveness', 'valence', 'length_duration', 'acousticness',
    'speechiness', 'popularity'
]

resumen_outliers = []

for col in cols_num:
    if col in df.columns:
        q1, q3 = df[col].quantile([0.25, 0.75])
        iqr = q3 - q1
        lim_inf = q1 - 1.5 * iqr
        lim_sup = q3 + 1.5 * iqr

        cant_outliers = df[(df[col] < lim_inf) | (df[col] > lim_sup)].shape[0]
        pct_outliers = round((cant_outliers / len(df)) * 100, 2)

        resumen_outliers.append({
            'columna': col,
            'limite_inferior': round(lim_inf, 2),
            'limite_superior': round(lim_sup, 2),
            'cant_outliers': cant_outliers,
            '%_outliers': pct_outliers
        })

df_outliers = pd.DataFrame(resumen_outliers)
display(df_outliers)
```

```python
import pandas as pd
import numpy as np

# Estandarizar nombres de columnas a snake_case
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
    .str.replace('(', '', regex=False)
    .str.replace(')', '', regex=False)
    .str.replace('%', 'pct')
)

# EXPLORACIÓN DE VARIABLES NUMÉRICAS
print("--- EXPLORACIÓN DE VARIABLES NUMÉRICAS ---")

# Detectar las columnas numéricas presentes en el dataframe
cols_num_deseadas = [
    'beats_per_minute_bpm', 'bpm', 'energy', 'danceability', 'loudness_db', 'dB',
    'liveness', 'valence', 'length_duration', 'length', 'acousticness',
    'speechiness', 'popularity', 'pop'
]

# Filtrar solo las columnas numéricas que existen en el df
cols_num = [c for c in df.columns if c in cols_num_deseadas or pd.api.types.is_numeric_dtype(df[c])]
# Excluir 'year' o 'index' de las métricas de audio si no corresponden
cols_num = [c for c in cols_num if c not in ['year', 'unnamed:_0', 'index', 'id']]

stats_list = []

for col in cols_num:
    s = df[col]
    q1 = s.quantile(0.25)
    q3 = s.quantile(0.75)
    iqr = q3 - q1

    stats_list.append({
        'variable': col,
        'media': round(s.mean(), 2),
        'mediana': round(s.median(), 2),
        'desvio_std': round(s.std(), 2),
        'iqr': round(iqr, 2),
        'sesgo_skew': round(s.skew(), 2),
        'min': s.min(),
        'p1': round(s.quantile(0.01), 2),
        'p99': round(s.quantile(0.99), 2),
        'max': s.max()
    })

df_stats_num = pd.DataFrame(stats_list)
display(df_stats_num)

# EXPLORACIÓN DE VARIABLES CATEGÓRICAS
print("\n--- EXPLORACIÓN DE VARIABLES CATEGÓRICAS ---")

# Identificar columnas de artista y género
col_artist = [c for c in df.columns if 'artist' in c][0] if any('artist' in c for c in df.columns) else 'artist'
col_genre = [c for c in df.columns if 'genre' in c][0] if any('genre' in c for c in df.columns) else 'top_genre'

cols_cat = [col_artist, col_genre]

# Cardinalidad
print("Cardinalidad (valores únicos por categoría):")
for col in cols_cat:
    if col in df.columns:
        unicos = df[col].nunique()
        print(f"- {col}: {unicos} categorías distintas")

# Frecuencias: Top 10 Artistas
if col_artist in df.columns:
    print("\nTop 10 Artistas con más canciones en el dataset:")
    display(df[col_artist].value_counts().head(10).reset_index())

# Frecuencias: Top 10 Géneros
if col_genre in df.columns:
    print("\nTop 10 Géneros dominantes:")
    display(df[col_genre].value_counts().head(10).reset_index())

    # Géneros raros (con solo 1 canción)
    generos_raros = (df[col_genre].value_counts() == 1).sum()
    print(f"\nCantidad de géneros con solo 1 canción registrada: {generos_raros}")
```

```python
import pandas as pd

# 1. Limpieza profunda de los nombres de columnas (elimina ';' y espacios)
df.columns = [
    c.replace(';', '').strip().lower().replace(' ', '_')
    for c in df.columns
]

# 2. Eliminar columnas vacías sobrantes (las que se crearon por los ';' al final)
df = df.loc[:, ~df.columns.str.contains('unnamed')].copy()

# 3. Limpieza y conversión de valores numéricos con ';'
for col in df.columns:
    if df[col].dtype == 'object':
        temp_cleaned = df[col].astype(str).str.replace(';', '').str.strip()
        if pd.to_numeric(temp_cleaned, errors='coerce').notna().sum() > 0.5 * len(df):
            df[col] = pd.to_numeric(temp_cleaned, errors='coerce')

# 4. Mapeo de variables clave
col_pop = 'popularity' if 'popularity' in df.columns else [c for c in df.columns if 'pop' in c][0]
col_year = 'year' if 'year' in df.columns else [c for c in df.columns if 'year' in c or 'año' in c][0]
col_genre = 'top_genre' if 'top_genre' in df.columns else [c for c in df.columns if 'genre' in c][0]
col_loud = 'loudness_db' if 'loudness_db' in df.columns else [c for c in df.columns if 'loud' in c or 'db' in c][0]
col_energy = 'energy' if 'energy' in df.columns else [c for c in df.columns if 'energ' in c][0]

print(f"Popularidad: '{col_pop}' (Tipo: {df[col_pop].dtype})")
print(f"Año: '{col_year}' (Tipo: {df[col_year].dtype})")
print(f"Género: '{col_genre}' (Tipo: {df[col_genre].dtype})")
print(f"Volumen: '{col_loud}' (Tipo: {df[col_loud].dtype})")
print(f"Energía: '{col_energy}' (Tipo: {df[col_energy].dtype})")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

sns.set_theme(style="whitegrid")
sns.set_palette("colorblind")

# GRAFICO 1: SERIE TEMPORAL CON LÍNEA DE REFERENCIA DE PROMEDIO (axhline)
popularidad_anio = df.groupby('year')['popularity'].mean().reset_index()

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(popularidad_anio['year'], popularidad_anio['popularity'], marker="o", color="#0173b2")

prom_global = popularidad_anio['popularity'].mean()
ax.axhline(prom_global, ls="--", color="red", alpha=0.7, label=f"Promedio ({prom_global:.1f})")

ax.set_title("Evolución anual de la popularidad promedio", fontweight='bold')
ax.set_xlabel("Año de lanzamiento")
ax.set_ylabel("Popularidad Promedio")
ax.legend()
plt.xticks(rotation=45)
plt.tight_layout()

fig = plt.gcf()
fig.savefig("evolucion_popularidad.png", dpi=150, bbox_inches="tight")
plt.show()

# GRAFICO 2: TOP CATEGORÍAS CON BARH
top_cat = df['top_genre'].value_counts().head(5).reset_index()
top_cat.columns = ['categoria', 'monto']
top_cat_sorted = top_cat.sort_values("monto", ascending=True)

plt.figure(figsize=(8, 4.5))
plt.barh(top_cat_sorted["categoria"], top_cat_sorted["monto"], color="#de8f5f")
plt.title("Top 5 géneros por cantidad de canciones", fontweight='bold')
plt.xlabel("Cantidad de canciones")
plt.ylabel("Género Musical")
plt.tight_layout()

fig = plt.gcf()
fig.savefig("top5_generos.png", dpi=150, bbox_inches="tight")
plt.show()

# GRAFICO 3: DISPERSIÓN CON MUESTRA Y LEYENDA (loudness vs energy)
muestra = df.sample(800, random_state=42) if len(df) > 800 else df

plt.figure(figsize=(7.5, 5))
sns.scatterplot(data=muestra, x="loudness_db", y="energy", hue="top_genre", alpha=0.7)
plt.title("Relación Volumen vs Energía por Género", fontweight='bold')
plt.xlabel("Volumen (Loudness dB)")
plt.ylabel("Energía (0 - 100)")
plt.legend(title="Género", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()

fig = plt.gcf()
fig.savefig("dispersion_volumen_energia.png", dpi=150, bbox_inches="tight")
plt.show()

# GRAFICO 4: HISTOGRAMA CON DENSIDAD (histplot + kde)
plt.figure(figsize=(7, 4))
sns.histplot(df["popularity"].dropna(), kde=True, bins=30, color="#029e73")
plt.title("Distribución de la Popularidad por Canción", fontweight='bold')
plt.xlabel("Popularidad (0 - 100)")
plt.ylabel("Frecuencia")
plt.tight_layout()

fig = plt.gcf()
fig.savefig("distribucion_popularidad.png", dpi=150, bbox_inches="tight")
plt.show()

# GRAFICO 5: BOXPLOT DE CAJA POR CATEGORÍA
top_5_genres = df['top_genre'].value_counts().head(5).index.tolist()

df_box = df[df['top_genre'].isin(top_5_genres)].copy()
df_box['top_genre'] = df_box['top_genre'].astype(str)

plt.figure(figsize=(9, 5))
sns.boxplot(
    data=df_box,
    x="top_genre",
    y="popularity",
    order=top_5_genres,
    palette="colorblind"
)

plt.title("Distribución de la Popularidad por Top 5 Géneros", fontweight='bold', fontsize=12)
plt.xlabel("Género Musical", fontweight='bold')
plt.ylabel("Popularidad (0 - 100)", fontweight='bold')
plt.xticks(rotation=15, ha='right')
plt.tight_layout()

fig = plt.gcf()
fig.savefig("boxplot_popularidad_genero_corregido.png", dpi=150, bbox_inches="tight")
plt.show()
```
